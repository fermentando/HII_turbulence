```python
import numpy as np
import matplotlib.pyplot as plt 
import powerlaw
```

```python
data = np.load('Figures/Sh2-311_NII_order2.npz')
vsf = data['vsf']
r = data['log_centers']
print(10**0.5)
mask = (r < np.log10(50) ) & (r > np.log10(5))

log_r = r[mask]
vsf_fit = vsf[mask]
log_vsf = np.log10(vsf_fit)

alpha, logC = np.polyfit(log_r, log_vsf, 1)

print("Power-law slope α =", alpha)

plt.loglog(10**r, vsf, 'o')
plt.loglog(10**r[mask], 10**logC * (10**r[mask])**alpha, 'r-', lw=2)
plt.xlabel('r')
plt.ylabel('vsf')
plt.show()

data4 = np.load('Figures/Sh2-311_NII_order4.npz')
vsf4 = data4['vsf']
r4 = data4['log_centers']
log_r = r4[mask]
vsf_fit = np.sqrt(vsf4[mask] - vsf[mask]**2)/vsf[mask]
log_vsf = np.log10(vsf_fit)

alpha, logC = np.polyfit(log_r, log_vsf, 1)
print("slope ratio α =", alpha)
plt.clf()
plt.loglog(10**r4, np.sqrt(vsf4 - vsf**2)/vsf, 'o')
plt.loglog(10**r[mask], 10**logC * (10**r[mask])**alpha, 'r-', lw=2)
plt.xlabel('r')
plt.ylabel('ratio')
plt.show()
```
